```python
import chromadb

client = chromadb.Client()

print("Client created!")
```

```python
practice_collection = client.get_or_create_collection(
    name="practice_collection"
)

print("Practice collection created!")
```

```python
practice_collection.add(
    documents=[
        "I enjoy learning Python.",
        "Artificial intelligence is changing technology.",
        "I like watching movies.",
        "Cricket is a popular sport.",
        "I am learning about databases."
    ],
    ids=["p1", "p2", "p3", "p4", "p5"]
)

print("5 documents added!")
```

```python
results = practice_collection.query(
    query_texts=["Tell me about programming"],
    n_results=2
)

print(results["documents"])
```

📝 Exercise 2

```python
metadata_collection = client.get_or_create_collection(
    name="metadata_practice"
)

print("Metadata collection created!")
```

```python
metadata_collection.add(
    documents=[
        "Python is used for software development.",
        "Artificial intelligence is part of modern technology.",
        "Banks use technology for secure transactions.",
        "Stock markets are part of the finance industry."
    ],
    metadatas=[
        {"category": "technology"},
        {"category": "technology"},
        {"category": "finance"},
        {"category": "finance"}
    ],
    ids=["m1", "m2", "m3", "m4"]
)

print("4 documents added!")
```

```python
results = metadata_collection.query(
    query_texts=["Tell me about technology"],
    n_results=2,
    where={"category": "technology"}
)

print(results["documents"])
```

```python
results = metadata_collection.query(
    query_texts=["Tell me about finance"],
    n_results=2,
    where={"category": "finance"}
)

print(results["documents"])
```

📝 Exercise 3

```python
from chromadb.utils import embedding_functions

mpnet_embedder = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-mpnet-base-v2"
)

print("MPNet model loaded!")
```

```python
mpnet_collection = client.get_or_create_collection(
    name="mpnet_collection",
    embedding_function=mpnet_embedder
)

print("MPNet collection created!")
```

```python
mpnet_collection.add(
    documents=[
        "Artificial intelligence and neural networks are important technologies.",
        "Machine learning allows computers to learn from data.",
        "Football is a popular sport played around the world."
    ],
    ids=["mp1", "mp2", "mp3"]
)

print("Documents added!")
```

```python
from chromadb.utils import embedding_functions

minilm_embedder = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-MiniLM-L6-v2"
)

print("MiniLM model loaded!")
```

```python
minilm_collection = client.get_or_create_collection(
    name="minilm_collection",
    embedding_function=minilm_embedder
)

print("MiniLM collection created!")
```

```python
minilm_collection.add(
    documents=[
        "Artificial intelligence and neural networks are important technologies.",
        "Machine learning allows computers to learn from data.",
        "Football is a popular sport played around the world."
    ],
    ids=["mini1", "mini2", "mini3"]
)

print("Documents added!")
```

```python
results = minilm_collection.query(
    query_texts=["Tell me about artificial intelligence"],
    n_results=2
)

print("Documents:")
print(results["documents"])

print("\nDistances:")
print(results["distances"])
```
