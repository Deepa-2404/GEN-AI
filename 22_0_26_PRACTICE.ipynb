```python
import chromadb

print("ChromaDB imported successfully!")
```

```python
client = chromadb.Client()

collection = client.create_collection(name="crud_demo")

print("Collection created:", collection.name)
print("Current document count:", collection.count())
```

```python
collection.add(
    documents=[
        "Python is easy to learn.",
        "I like watching cricket.",
        "The sky is blue.",
        "My favorite food is pizza.",
        "JavaScript is used for web development."
    ],
    metadatas=[
        {"category": "tech"},
        {"category": "sports"},
        {"category": "nature"},
        {"category": "food"},
        {"category": "tech"}
    ],
    ids=["doc7", "doc8", "doc9", "doc10", "doc11"]
)

print("Total count:", collection.count())
```

```python
collection.upsert(
    documents=[
        "Python is easy and powerful to learn.",
        "I like watching cricket and football.",
        "ChromaDB is a vector database."
    ],
    ids=["doc7", "doc8", "doc12"]
)

print("Total count after upsert:", collection.count())
```

```python
print(collection.get(ids=["doc7", "doc8", "doc12"]))
```

```python
collection.delete(
    where={"category": "food"}
)

print("Total count after category delete:", collection.count())
print("Remaining ids:", collection.get()["ids"])
```

```python
def safe_add(collection, id, text):
    collection.upsert(
        ids=[id],
        documents=[text]
    )
```
